# Import


```python
import torch

import torch.nn as nn
```

# Data

```python
vocab = {
    "I" : 0,
    "Love" : 1,
    "AI" : 2,
    "and" : 3,
    "PyTorch" : 4
}

vocab_size = len(vocab)
vocab_size
```

# Convert sentence to tokens

```python
# I Love AI  -> 012
# just a lookup table

tokens = torch.tensor([0, 1, 2])

print(tokens)
print(tokens.shape)
```

# Create Embedding Layer

```python
embedding_dim = 8

# creating a obj for embedding
# text to embed Example:
# I -> [-0.6207,  0.0349,  0.0896, -1.3563, -0.9755,  1.4167, -1.8134, -2.1788]
# Why 8? cause we put embedding_dim = 8

embedding = nn.Embedding(
    num_embeddings = vocab_size,
    embedding_dim = embedding_dim
)
```

```python
# in tokens i have only 3 (0, 1, 2)
# just accessing embedding for the first 3 tokens

# illustration
# I    -> 0 -> [-0.6207,  0.0349,  0.0896, -1.3563, -0.9755,  1.4167, -1.8134, -2.1788]
# Love -> 1 -> [ 0.0684,  0.0784, -0.8352,  1.1685,  0.6589, -0.7665, -0.6419,  0.0317]
# AI   -> 2 -> [ 0.4224,  0.7009, -0.9845, -1.1494,  1.1792,  0.7312, -0.8205, -0.2286]

x = embedding(tokens)
x
```

```python
x.shape
```

# Adding Linear Layer

```python
# create a obj for Linear

linear = nn.Linear(
    in_features = 8,
    out_features = 8
)
```

# use the x in Linear Layer

```python
y = linear(x)

print("Tokens Shape:", tokens.shape)
print("Embedding Shape:", x.shape)
print("Output Shape:", y.shape)
```
