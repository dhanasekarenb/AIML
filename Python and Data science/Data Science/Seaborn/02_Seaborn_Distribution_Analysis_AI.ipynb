# Why Distribution Analysis Matters

Before training a model, ask:

- Is the data normal?
- Is the data skewed?
- Are there outliers?
- Does the feature need scaling?

Distribution analysis answers these questions.


````python

import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
````


# What is a Distribution?

A distribution describes how values are spread.

Example:

```python
[10, 11, 12, 13, 14]
```

Very concentrated.

Example:

```python
[1, 50, 100, 150, 200]
```

Highly spread out.



# Histplot

The most common distribution plot.

Histograms divide data into bins and count observations.

### AI Usage

- Understand feature distributions
- Detect skewness
- Detect unusual values


````python
data = np.random.randn(1000)

sns.histplot(data)

plt.title("Histogram")

plt.show()
````


# Understanding Bins

Bins group values into ranges.

More bins:

- More detail

Fewer bins:

- Simpler view


````python

data = np.random.randn(1000)

sns.histplot(
    data,
    bins=10
)

plt.show()
````

````python

data = np.random.randn(1000)

sns.histplot(
    data,
    bins=50
)

plt.show()
````


# KDE Plot

KDE = Kernel Density Estimation

Think of it as a smooth histogram.

Instead of bars, we get a smooth curve.


````python

data = np.random.randn(1000)

sns.kdeplot(data)

plt.title("KDE Plot")

plt.show()
````


# Why AI Engineers Use KDE

KDE helps us:

- Understand shape
- Compare distributions
- Detect multimodal data


````python

group1 = np.random.normal(0,1,1000)
group2 = np.random.normal(2,1,1000)

sns.kdeplot(group1, label="Group 1")
sns.kdeplot(group2, label="Group 2")

plt.legend()

plt.show()
````


# Histplot + KDE

A very common combination.


````python

data = np.random.randn(1000)

sns.histplot(
    data,
    kde=True
)

plt.show()
````


# Normal Distribution

Most famous statistical distribution.

Characteristics:

- Bell shaped
- Symmetric
- Mean near center


````python

normal = np.random.normal(
    loc=0,
    scale=1,
    size=5000
)

sns.histplot(
    normal,
    kde=True
)

plt.title("Normal Distribution")

plt.show()
````


# Right Skewed Distribution

Common examples:

- Income
- Salary
- House Prices


````python
right_skewed = np.random.exponential(
    scale=1,
    size=5000
)

sns.histplot(
    right_skewed,
    kde=True
)

plt.title("Right Skewed")

plt.show()
````


# Left Skewed Distribution


````python

left_skewed = -np.random.exponential(
    scale=1,
    size=5000
)

sns.histplot(
    left_skewed,
    kde=True
)

plt.title("Left Skewed")

plt.show()
````


# Uniform Distribution

All values have similar probability.


````python

uniform = np.random.uniform(
    0,
    10,
    5000
)

sns.histplot(
    uniform,
    kde=True
)

plt.title("Uniform Distribution")

plt.show()
````


# Displot

Figure-level distribution plot.

Can generate:

- Histograms
- KDE
- Multiple distributions


````python

tips = sns.load_dataset("tips")

sns.displot(
    tips,
    x="total_bill"
)
````


# Rug Plot

Shows individual observations.

Useful for understanding density.


````python

data = np.random.randn(100)

sns.rugplot(data)

plt.show()
````


# Comparing Distributions

A common EDA task.


````python

tips = sns.load_dataset("tips")

sns.histplot(
    data=tips,
    x="total_bill",
    hue="sex",
    kde=True
)

plt.show()
````


# AI Connection — Why Scaling Exists

Suppose:

```python
Age      : 20-60
Salary   : 20,000-100,000
```

Different scales create problems.

Distribution analysis helps decide:

- StandardScaler
