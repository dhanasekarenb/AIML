````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
````

## Why Correlation Matters

Before training a model we ask:

- Which features are related?
- Which features affect the target?
- Which features are redundant?
- Which features should be removed?

Correlation analysis helps answer these questions.


## What is Correlation?

Correlation measures how strongly two variables move together.

Values range from:

```text
-1  → Perfect Negative Correlation
 0  → No Correlation
+1  → Perfect Positive Correlation
```


## Positive Correlation

As one variable increases, the other also increases.

Example:

- Experience vs Salary
- Study Hours vs Marks


## Negative Correlation

As one variable increases, the other decreases.

Example:

- TV Hours vs Exam Score
- Price vs Demand


## No Correlation

Variables move independently.

Knowing one variable tells us little about the other.


## Correlation Matrix

A correlation matrix stores correlations between every feature pair.


````python
data = pd.DataFrame({
    "Age":[20,25,30,35,40],
    "Salary":[25000,35000,45000,55000,65000],
    "Experience":[1,3,5,7,9],
    "gaming":[10, 8, 6, 2, 1]
})

data.corr()
````

## Heatmap

Heatmaps visualize the correlation matrix using colors.


````python
corr = data.corr()

sns.heatmap(corr)

plt.title("Correlation Heatmap")

plt.show()
````

## Understanding Heatmap Colors

Typical interpretation:

- Bright colors → Strong relationship
- Dark colors → Weak relationship

Always read the values, not just the colors.


## Annotated Heatmaps

Annotations display correlation values directly.


````python
corr = data.corr()

sns.heatmap(
    corr,
    annot=True
)

plt.show()
````

## Why AI Engineers Use Annotated Heatmaps

The exact correlation value matters.

Examples:

```text
0.95 → Very Strong
0.30 → Weak
```


## Using Color Maps

Color maps improve readability.


````python
corr = data.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
````

## Real Dataset Example

Using the Iris dataset.


````python
from  sklearn.datasets import load_iris

iris_raw = load_iris(as_frame=True)

iris = iris_raw.frame.copy()

iris.columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "target"
]

iris["species"] = iris["target"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

iris = iris.drop(columns="target")

corr = iris.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Iris Correlation Matrix")

plt.show()
````

## Multicollinearity

Multicollinearity occurs when features are highly correlated.

Example:

```text
Age
Birth Year
```

Both contain nearly the same information.


## Why Multicollinearity Matters

Problems:

- Unstable coefficients
- Harder interpretation
- Reduced model quality

Especially important for:

- Linear Regression
- Logistic Regression


## Feature Selection Using Correlation

Questions:

- Which features are strongly related to the target?
- Which features are redundant?
- Which features should be removed?


````python
data = pd.DataFrame({
    "StudyHours":[1,2,3,4,5],
    "Marks":[40,50,65,80,90],
    "TVHours":[5,4,3,2,1]
})

corr = data.corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.show()
````
