## Your First Plot

````python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,30,40]

plt.plot(x,y)
plt.show()
````

## Labels and Titles

````python
import matplotlib.pyplot as plt

plt.plot([1,2,3],[10,20,30])

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")

plt.show()
````

## Line Styles

````python
plt.plot([1,2,3],[4,5,6], linestyle="--")
plt.show()
````

## Colors

````python
plt.plot([1,2,3],[4,5,6], color="red")
plt.show()
````

## Markers

````python
plt.plot([1,2,3],[4,5,6], marker="o")
plt.show()
````

## Grid

````python
plt.plot([1,2,3],[4,5,6])
plt.grid(True)
plt.show()
````

## Legend

````python
plt.plot([1,2,3],[4,5,6], label="Model A")
plt.plot([1,2,3],[3,4,8], label="Model B")

plt.legend()
plt.show()
````

## Figure Size

````python
plt.figure(figsize=(10,5))
plt.plot([1,2,3],[4,5,6])
plt.show()
````

## Saving Figures

````python
plt.plot([1,2,3],[4,5,6])
plt.savefig("my_plot.png")
````

## Mini Project — Training Loss Curve

````python
import matplotlib.pyplot as plt

epochs = [1,2,3,4,5,6,7]
loss = [2.4,1.9,1.4,1.1,0.8,0.6,0.5]

plt.figure(figsize=(8,5))
plt.plot(epochs, loss, marker="o", label="Training Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss Curve")

plt.grid(True)
plt.legend()

plt.show()
````
