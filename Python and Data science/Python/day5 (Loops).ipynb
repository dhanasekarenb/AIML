# For Loop
##### A for loop in Python is used to iterate over a sequence (such as a list, tuple, dictionary, set, or string) and execute a block of code for each element in the sequence.

```python
a="Orange"
for i in a:
    print(i)
```

```python
for i in "Apple":
    print(i)
```

```python
#List 
a=[1,2,3,4,5]
for i in a:
    print(i)
```

```python
fruits=["apple","banana","cherry"]
for i in fruits:
    print(i)
```

```python
a="banana"
for i in range(len(a)):
    # print(a[i])
    print(i)
```

```python
for i in range(1,11):
    print(i)
```

```python
for i in range(2,10,2):
    print(i)
```

```python
#Example of even numbers from 0 to 10

for i in range(10):
    if i%2==0:
        print(i)
```

#### LOOPING CONTROL STATEMENT

##### BREAK -- Stop looping when a certain condition is met.
##### CONTINUE -- Skip certain values without breaking the loop.

```python
#Break statement
for i in range(1,11):
    if i==6:
        break
    print(i)
```

```python
#Continue statement
for i in range(1,11):
    if i==6:
        continue
    print(i)
```

#### Nested For Loop

```python
for i in range(3):
    for j in range(3):
        print(i,j)
```

```python
for i in range(5,0,-1):
    print(i)
```

```python
#print pyramid
n=5
# Outer Loop
#range(n) generates numbers from 0 to n-1 → 0, 1, 2, 3, 4
for i in range(n):
    # Inner Loop
    #For each i, range(i + 1) generates numbers from 0 to i.
    #This means:
    #When i = 0 → range(1) → j = 0 → prints 1 star
    #When i = 1 → range(2) → j = 0, 1 → prints 2 stars
    #When i = 2 → range(3) → j = 0, 1, 2 → prints 3 stars
    #… and so on.
    for j in range(i+1):
        print("*",end="")
        print()
```

```python
#reverse pyramid
n=5
for i in range(n,0,-1):
    for j in range(i):
        print("*",end="")
    print()
```

```python
n=5
for i in range(n):
    for j in range(n):
        print("*",end="")
    print()
    j-=1
```

# WHILE LOOP
##### A while loop repeatedly executes a block of code as long as a given condition is True

```python
while True:
    print("hello")
    break
```

```python
i=0
while True:
    i+=1
    if i==5:
        break
    print("hello")
```

```python
i=1
while i<=5:
    print(i)
    i+=1
```

```python
#using break and continue in while loop
i=0
while i<10:
    
    # if i==5:
    #     break
    
    print(i)
    i+=1
```

```python
a=[1,2,3,4,5]
i=len(a)
while i>=0:
    print(a[i])
    i-=1
```

```python
i=0
while i<10:
    i+=1
    if i==5:
        continue
    
    print(i)
```
