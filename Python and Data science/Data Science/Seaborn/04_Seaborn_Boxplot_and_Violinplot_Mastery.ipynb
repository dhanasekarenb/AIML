```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Understanding Quartiles

Quartiles divide data into four equal parts.

- Q1 = 25%
- Q2 = Median
- Q3 = 75%

These are the foundation of boxplots.


## Interquartile Range (IQR)

Formula:

IQR = Q3 - Q1

Used for:

- Measuring spread
- Detecting outliers


## What is a Boxplot?

A boxplot summarizes a distribution using:

- Minimum
- Q1
- Median
- Q3
- Maximum

Potential outliers are shown separately.


```python
tips = sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x="total_bill"
)

plt.title("Boxplot of Total Bill")
plt.show()
```

## Detecting Outliers

Outliers appear outside the whiskers.

These points deserve investigation.

They may be:

- Data entry errors
- Rare events
- Valuable business insights


```python
data = [10,12,13,14,15,16,18,20,100]

sns.boxplot(x=data)

plt.title("Outlier Example")
plt.show()
```

## Comparing Categories with Boxplots

One of the most common EDA tasks.


```python
tips = sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Bills by Day")
plt.show()
```

## Using Hue with Boxplots

Compare multiple groups simultaneously.


```python
tips = sns.load_dataset("tips")

sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex"
)

plt.show()
```

## What is a Violin Plot?

A violin plot combines:

- Boxplot
- Density estimation

It shows both spread and shape.


```python
tips = sns.load_dataset("tips")

sns.violinplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Violin Plot")
plt.show()
```

## Why Violin Plots Are Powerful

A boxplot shows summary statistics.

A violin plot shows:

- Density
- Shape
- Multiple peaks
- Distribution patterns


```python
tips = sns.load_dataset("tips")

sns.violinplot(
    data=tips,
    x="sex",
    y="tip"
)

plt.show()
```

## Boxplot vs Violinplot

Boxplot:
- Faster to read
- Simpler
- Great for reports

Violinplot:
- More information
- Better for analysis
- Shows density


## What is a Boxenplot?

Boxenplot is designed for large datasets.

Advantages:

- Better tail analysis
- Better for large samples
- More detailed than boxplots


```python
tips = sns.load_dataset("tips")

sns.boxenplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
```
