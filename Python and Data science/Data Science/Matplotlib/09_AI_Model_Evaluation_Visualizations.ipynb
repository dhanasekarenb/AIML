# Confusion Matrix

A confusion matrix shows:

- True Positives (TP)
- True Negatives (TN)
- False Positives (FP)
- False Negatives (FN)

This is one of the most important ML visualizations.


````python

import matplotlib.pyplot as plt
import numpy as np

cm = np.array([
    [90, 10],
    [5, 95]
])

plt.imshow(cm)

plt.title("Confusion Matrix")
plt.colorbar()

plt.show()
````


# Reading a Confusion Matrix

Example:

```text
90  10
5   95
```

Interpretation:

- 90 correct negatives
- 95 correct positives
- 10 false positives
- 5 false negatives
```

# ROC Curve

ROC = Receiver Operating Characteristic

Measures:

True Positive Rate

vs

False Positive Rate


````python

import numpy as np
import matplotlib.pyplot as plt

fpr = [0,0.1,0.2,0.4,1]
tpr = [0,0.5,0.75,0.9,1]

plt.plot(fpr, tpr)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.title("ROC Curve")

plt.show()
````


# Understanding ROC Curves

Better models push closer to the top-left corner.

Ideal model:

```text
TPR = 1
FPR = 0
```



# AUC

AUC = Area Under Curve

Interpretation:

```text
1.0  -> Perfect
0.9  -> Excellent
0.8  -> Good
0.5  -> Random Guessing
```



# Precision

Question:

Of all positive predictions,

How many were actually correct?



# Recall

Question:

Of all actual positives,

How many did we find?



# Precision–Recall Curve

Important when datasets are imbalanced.

Examples:

- Fraud Detection
- Disease Detection
- Defect Detection


````python

import matplotlib.pyplot as plt

recall = [0,0.2,0.4,0.6,0.8,1]
precision = [1,0.95,0.9,0.85,0.7,0.5]

plt.plot(recall, precision)

plt.xlabel("Recall")
plt.ylabel("Precision")

plt.title("Precision Recall Curve")

plt.show()
````


# Learning Curves

One of the most useful deep learning visualizations.

Used to monitor:

- Training Loss
- Validation Loss
- Overfitting
- Underfitting


````python

import matplotlib.pyplot as plt

epochs = [1,2,3,4,5,6]

train_loss = [2.5,1.8,1.2,0.8,0.5,0.3]
val_loss = [2.6,2.0,1.4,1.2,1.5,1.9]

plt.plot(epochs, train_loss, label="Train")
plt.plot(epochs, val_loss, label="Validation")

plt.legend()

plt.title("Learning Curve")

plt.show()
````


# Detecting Overfitting

Signs:

- Training loss decreases
- Validation loss increases

Model memorizes training data.



# Detecting Underfitting

Signs:

- Both losses remain high
- Model cannot learn patterns



# Validation Curves

Used for hyperparameter tuning.

Example:

```python
max_depth
learning_rate
n_estimators
```


````python

import matplotlib.pyplot as plt

depth = [1,2,3,4,5,6]
accuracy = [70,78,85,90,88,84]

plt.plot(depth, accuracy)

plt.xlabel("Max Depth")
plt.ylabel("Accuracy")

plt.title("Validation Curve")

plt.show()
````


# Comparing Multiple Models

Very common in real projects.


````python

import matplotlib.pyplot as plt

models = ["RF","XGB","SVM"]
accuracy = [89,94,87]

plt.bar(models, accuracy)

plt.title("Model Comparison")

plt.show()
````


# Professional Evaluation Dashboard

Combining multiple visualizations together.


````python

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(
    2,
    2,
    figsize=(10,8)
)

ax[0,0].plot([1,2,3],[2,1,0.5])
ax[0,0].set_title("Loss")

ax[0,1].plot([1,2,3],[60,75,90])
ax[0,1].set_title("Accuracy")

ax[1,0].imshow(np.array([[90,10],[5,95]]))
ax[1,0].set_title("Confusion Matrix")

ax[1,1].bar(
    ["RF","XGB","SVM"],
    [89,94,87]
)

ax[1,1].set_title("Models")

plt.tight_layout()
plt.show()
````
