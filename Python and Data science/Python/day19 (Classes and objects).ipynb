# Classes and Objects

## What is a Class?
A **Class** is a **blueprint** or **template** used to create objects.

**Class = Blueprint**

Example:
- Student
- Car
- Employee

## What is an Object?
An **Object** is an **instance** of a class.

It contains the actual data.

## Real World Analogy
| Class | Object |
|---|---|
| Student | Arun |
| Car | Tesla |
| Mobile | iPhone |
| Book | Python Book |

```python
class Student:
    pass

student1 = Student()
print(student1)
```

## Attributes
Attributes are variables that belong to an object.

```python
class Student:
    pass

student = Student()
student.name = "Arun"
student.age = 20
student.department = "CSE"

print(student.name)
print(student.age)
print(student.department)
```

## Multiple Objects
One class can create multiple objects. Each object stores different values.

```python
class Student:
    pass

s1 = Student()
s2 = Student()

s1.name = "Arun"
s2.name = "Bala"

print(s1.name)
print(s2.name)
```

## Example Program

```python
class Student:
    pass

student = Student()
student.name = "Dhana"
student.roll_no = 101
student.department = "AI"
student.mark = 92

print("Name:", student.name)
print("Roll No:", student.roll_no)
print("Department:", student.department)
print("Mark:", student.mark)
```

```python
class Student:
    def __init__(self, name, roll_no, department, mark):
        self.name = name
        self.roll_no = roll_no
        self.department = department
        self.mark = mark

student = Student("Dhanasekaren", "005", "CSE",  98)

print("Name:", student.name)
print("Roll No:", student.roll_no)
print("Department:", student.department)
print("Mark:", student.mark)
```

## Why Use Classes?
- Organize related data
- Reuse code
- Create many similar objects
- Makes programs easier to maintain

## Important Points
- Class is a blueprint.
- Object is an instance.
- One class can create many objects.
- Attributes store object data.

## Summary
- **Class → Blueprint**
- **Object → Instance**
- **Attribute → Variable inside an object**
