## 1. What is PyTorch?

**PyTorch** is an open-source machine learning and deep learning framework developed primarily by Meta AI.

It provides tools for:

- Tensor computation
- Automatic differentiation
- Neural networks
- GPU acceleration
- Model training and inference
- Computer vision
- Natural language processing
- Reinforcement learning

The most fundamental object in PyTorch is the **tensor**.

### Tensor

A tensor is a generalization of:

- Scalar → 0-dimensional tensor
- Vector → 1-dimensional tensor
- Matrix → 2-dimensional tensor
- Higher-dimensional arrays → 3D, 4D, 5D, ...

A neural network ultimately performs mathematical operations on tensors.

````python
import torch

print("PyTorch version:", torch.__version__)
````

## 2. Why PyTorch instead of normal Python lists or NumPy?

Python lists are excellent for general programming, and NumPy is excellent for numerical computing.

PyTorch adds several features important for deep learning:

| Feature | Python List | NumPy | PyTorch |
|---|---:|---:|---:|
| Numerical operations | Limited | ✅ | ✅ |
| Multidimensional arrays | Limited | ✅ | ✅ |
| GPU acceleration | ❌ | Usually ❌ | ✅ |
| Automatic gradients | ❌ | ❌ | ✅ |
| Deep-learning ecosystem | ❌ | ❌ | ✅ |

PyTorch tensors are conceptually similar to NumPy arrays, but PyTorch is designed to work naturally with neural-network training and accelerators such as GPUs.

## 3.1 `torch.tensor()`

### Definition

`torch.tensor()` creates a tensor from existing data.

### Basic syntax

```python
torch.tensor(data, dtype=None, device=None, requires_grad=False)
```

### Important parameters

- `data` → the input values.
- `dtype` → numerical data type.
- `device` → CPU or GPU.
- `requires_grad` → whether PyTorch should track operations for gradients.

The most necessary parameter is **`data`**. The others are optional for simple tensor creation.

````python
x = torch.tensor([1, 2, 3, 4])

print(x)
print("Type:", type(x))
````

## 3.2 Tensor from nested lists

Nested lists create multidimensional tensors.

````python
matrix = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print(matrix)
````

Here:

- Number of rows = `2`
- Number of columns = `3`
- Shape = `(2, 3)`
- Dimensions = `2`

This distinction between **shape** and **number of dimensions** is extremely important in deep learning.

# 4. Tensor Shape and Dimensions

A tensor has several useful properties in other words we say attribute (which is simply a varible) and methods (simply functions)

- `shape`
- `size()`
- `ndim`
- `numel()`
- `dtype`
- `device`

````python
x = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("Tensor:")
print(x)

print("\nShape:", x.shape) # x.shape is just an alias for x.size()
print("Size:", x.size())
print("Dimensions:", x.ndim) # Dimensions
print("Number of elements:", x.numel()) # number of element in short numel
print("Data type:", x.dtype) # data type
print("Device:", x.device) # GPU or CPU
````

### Understanding the output

For a `(2, 3)` tensor:

```text
2 × 3 = 6 elements
```

So:

```python
x.shape   # torch.Size([2, 3])
x.ndim    # 2
x.numel() # 6
```

### Important idea

**Shape tells you the size of every dimension.**

For example:

```text
(3,)        → vector
(2, 3)      → matrix
(2, 3, 4)   → 3-dimensional tensor
(32, 3, 224, 224) → batch of RGB images
```

# 5. Data Types (`dtype`)

PyTorch supports many numerical data types.

Common ones include:

- `torch.float32`
- `torch.float64`
- `torch.float16`
- `torch.bfloat16`
- `torch.int8`
- `torch.int16`
- `torch.int32`
- `torch.int64`
- `torch.bool`

For many neural-network operations, **`float32`** is the standard starting point.

Class labels are commonly stored as **`int64` (`torch.long`)**, especially when using `CrossEntropyLoss`.

````python
x = torch.tensor([1, 2, 3])

print(x)
print(x.dtype)

y = torch.tensor([1.0, 2.0, 3.0])

print(y)
print(y.dtype)
````

## 5.1 Explicitly specifying `dtype`

The `dtype` parameter controls how values are represented in memory and how numerical operations are performed.

````python
x = torch.tensor([1, 2, 3], dtype=torch.float32)
y = torch.tensor([1, 2, 3], dtype=torch.float64)

print("x:", x)
print("x dtype:", x.dtype)

print("\ny:", y)
print("y dtype:", y.dtype)
````

### Why does `dtype` matter?

Changing dtype can affect:

- Memory usage
- Numerical precision
- Speed
- GPU compatibility
- Training performance

For example, `float64` generally uses more memory than `float32`.

Do not assume that "higher precision" automatically means "better" for deep learning. The appropriate dtype depends on the operation and hardware.

# 6. Creating Special Tensors

## `torch.zeros()`

Creates a tensor filled with zeros.

```python
torch.zeros(size, dtype=None, device=None, requires_grad=False)
```

The **size/shape** is the necessary input.

````python
x = torch.zeros(2, 3)

print(x)
print("Shape:", x.shape)
````

### What happens if we change the dimensions?

````python
x = torch.zeros(4, 2)

print(x)
print("Shape:", x.shape)
````

The output changed because we changed the requested shape:

```text
torch.zeros(2, 3) → 2 rows × 3 columns
torch.zeros(4, 2) → 4 rows × 2 columns
```

## `torch.ones()`

Creates a tensor filled with ones.

```python
torch.ones(size, dtype=None, device=None, requires_grad=False)
```

````python
x = torch.ones(3, 4)

print(x)
````

## `torch.full()`

Creates a tensor filled with a specified value.

```python
torch.full(size, fill_value, dtype=None, device=None, requires_grad=False)
```

Two important inputs are:

- `size` → tensor shape
- `fill_value` → value used for every element

````python
x = torch.full((2, 3), fill_value=7)

print(x)
````

### Change `fill_value`

````python
x = torch.full((2, 3), fill_value=100)

print(x)
````

Only the stored value changed because the **shape remained `(2, 3)`** while `fill_value` changed from `7` to `100`.

# 7. `torch.empty()`

`torch.empty()` creates a tensor without initializing its values.

```python
torch.empty(size, dtype=None, device=None, requires_grad=False)
```

**Important:** the values are not guaranteed to be zero.

The memory contains whatever values happened to exist in that memory location.

Therefore, use `torch.zeros()` when you specifically need zeros.

````python
x = torch.empty(2, 3)

print(x)
````

The values printed by `torch.empty()` should **not** be interpreted as meaningful data.

This function is useful when you plan to fill the tensor immediately and want to avoid the extra initialization step.

# 8. Creating Sequences

## `torch.arange()`

Creates values with a regular step.

```python
torch.arange(start=0, end, step=1, ...)
```

Important parameters:

- `start`
- `end`
- `step`

The `end` value is generally **exclusive**, similar to Python's `range()`.

````python
x = torch.arange(0, 10, 1)

print(x)
````

### Change the step

````python
x = torch.arange(0, 10, 2)

print(x)
````

The output changes because `step=2` tells PyTorch to increase each value by 2:

```text
0, 2, 4, 6, 8
```

## `torch.linspace()`

Creates a specified number of evenly spaced values between two endpoints.

```python
torch.linspace(start, end, steps)
```

Unlike `arange()`, you specify the **number of values** using `steps`.

````python
x = torch.linspace(0, 1, steps=5)

print(x)
````

### Change `steps`

````python
x = torch.linspace(0, 1, steps=11)

print(x)
````

Here the interval remains `0 → 1`, but the number of points changes.

This is useful when you want a fixed number of evenly spaced samples.

# 9. Random Tensors

Random tensors are heavily used when initializing neural-network parameters.

Common functions:

- `torch.rand()` → uniform distribution from approximately `[0, 1)`
- `torch.randn()` → standard normal distribution
- `torch.randint()` → random integers

## `torch.rand()`

```python
torch.rand(size, ...)
```

The size is the necessary input.

````python
x = torch.rand(2, 3)

print(x)
````

Every time you execute the cell, the values normally change because they are randomly generated.

````python
x = torch.rand(2, 3)

print(x)
````

## `torch.randn()`

`randn()` samples from a standard normal distribution with approximately:

```text
mean = 0
standard deviation = 1
```

````python
x = torch.randn(2, 3)

print(x)
````

### `rand()` vs `randn()`

```text
rand()  → uniform random values
randn() → normally distributed random values
```

The shape parameter controls the number of generated values, while the underlying function controls their distribution.

## `torch.randint()`

Creates random integer values.

```python
torch.randint(low, high, size)
```

Important parameters:

- `low` → minimum inclusive value
- `high` → maximum exclusive value
- `size` → output shape

````python
x = torch.randint(0, 10, (2, 5))

print(x)
````

The generated values can be:

```text
0, 1, 2, ..., 9
```

but not `10`, because `high=10` is exclusive.

# 10. Indexing

PyTorch indexing is similar to NumPy and Python.

````python
x = torch.tensor([10, 20, 30, 40, 50])

print("First element:", x[0])
print("Third element:", x[2])
print("Last element:", x[-1])
````

## Indexing a matrix

````python
x = torch.tensor([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("First row:", x[0])
print("Second row:", x[1])
print("Element at row 2, column 3:", x[1, 2])
````

```text
x[0, 0] → first row, first column
```

# 11. Slicing

You can select ranges of tensor elements.

```python
tensor[start:end]
```

The `end` index is exclusive.

````python
x = torch.tensor([10, 20, 30, 40, 50])

print(x[1:4])
print(x[:3])
print(x[2:])
print(x[::2])
````

### Understanding:

```text
x[1:4]  → indices 1, 2, 3
x[:3]   → indices 0, 1, 2
x[2:]   → index 2 onward
x[::2]  → every second element
```

# 12. Basic Tensor Arithmetic

PyTorch supports element-wise arithmetic.

Operations include:

- Addition `+`
- Subtraction `-`
- Multiplication `*`
- Division `/`
- Power `**`

````python
a = torch.tensor([1., 2., 3.])
b = torch.tensor([10., 20., 30.])

print("Addition:", a + b)
print("Subtraction:", b - a)
print("Multiplication:", a * b)
print("Division:", b / a)
print("Power:", a ** 2)
````

## Important: `*` vs matrix multiplication

In PyTorch:

```python
a * b
```

means **element-wise multiplication**.

For matrix multiplication:

```python
a @ b
```

or:

```python
torch.matmul(a, b)
```

This distinction becomes critical when building neural networks.

````python
a = torch.tensor([
    [1., 2.],
    [3., 4.]
])

b = torch.tensor([
    [5., 6.],
    [7., 8.]
])

print("Element-wise multiplication:")
print(a * b)

print("\nMatrix multiplication:")
print(a @ b)
````

The outputs differ because the operations perform different mathematics.

### Element-wise multiplication

```text
[1×5, 2×6]
[3×7, 4×8]
```

### Matrix multiplication

Each output element is calculated using a row-column dot product.

# 13. Reduction Operations

Reduction operations summarize tensor values.

Important examples:

- `sum()`
- `mean()`
- `min()`
- `max()`
- `std()`

````python
x = torch.tensor([1., 2., 3., 4., 5.])

print("Sum:", x.sum())
print("Mean:", x.mean())
print("Minimum:", x.min())
print("Maximum:", x.max())
print("Standard deviation:", x.std())
````

## Reduction along a dimension

For a matrix, you can specify `dim`.

```python
x.sum(dim=0)
x.sum(dim=1)
```

This is extremely important when working with batches.

````python
x = torch.tensor([
    [1., 2., 3.],
    [4., 5., 6.]
])

print("Tensor:")
print(x)

print("\nSum along dim=0:")
print(x.sum(dim=0))

print("\nSum along dim=1:")
print(x.sum(dim=1))
````

For shape `(2, 3)`:

- `dim=0` reduces the rows → one value per column.
- `dim=1` reduces the columns → one value per row.

Understanding dimensions is one of the most important skills in PyTorch.

# 14. Reshaping Tensors

Common operations:

- `reshape()`
- `view()`
- `flatten()`
- `squeeze()`
- `unsqueeze()`

## `reshape()`

Changes the shape while preserving the number of elements.

```python
x.reshape(new_shape)
```

The new shape must contain the same total number of elements.

````python
x = torch.arange(1, 7)

print("Original:")
print(x)
print("Shape:", x.shape)

y = x.reshape(2, 3)

print("\nReshaped:")
print(y)
print("Shape:", y.shape)
````

The original tensor has:

```text
6 elements
```

The new shape also contains:

```text
2 × 3 = 6 elements
```

So the reshape is valid.

Try changing `(2, 3)` to `(3, 2)`.

````python
x = torch.arange(1, 7)
y = x.reshape(3, 2)

print(y)
````

### Invalid reshape

Try this:

```python
x.reshape(4, 2)
```

It should fail because:

```text
original elements = 6
requested elements = 4 × 2 = 8
```

The number of elements cannot change during reshape.

````python
x = torch.arange(1, 7)

try:
    y = x.reshape(4, 2)
except RuntimeError as e:
    print("Expected error:")
    print(e)
````

# 15. `flatten()`

`flatten()` converts multiple dimensions into a single dimension.

This is commonly used before feeding image features into a fully connected layer.

````python
x = torch.tensor([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("Original shape:", x.shape)

y = x.flatten()

print("Flattened:", y)
print("Flattened shape:", y.shape)
````

The tensor has shape:

```text
(2, 2, 2)
```

After flattening:

```text
(8,)
```

because:

```text
2 × 2 × 2 = 8
```

# 16. `unsqueeze()` and `squeeze()`

## `unsqueeze(dim)`

Adds a dimension of size 1.

## `squeeze(dim)`

Removes dimensions whose size is 1.

````python
x = torch.tensor([1, 2, 3])

print("Original:", x.shape)

y = x.unsqueeze(0)
print("unsqueeze(0):", y.shape)

z = x.unsqueeze(1)
print("unsqueeze(1):", z.shape)
````

The position of `dim` determines **where** the new dimension is inserted.

This becomes important for image and batch dimensions.

For example, an image might have:

```text
(C, H, W)
```

and adding a batch dimension produces:

```text
(1, C, H, W)
```

````python
image = torch.rand(3, 224, 224)

print("Single image:", image.shape)

batch = image.unsqueeze(0)

print("With batch dimension:", batch.shape)
````

# 17. Concatenation and Stacking

## `torch.cat()`

Joins tensors along an existing dimension.

## `torch.stack()`

Creates a **new dimension** and places tensors along it.

````python
a = torch.tensor([1, 2])
b = torch.tensor([3, 4])

print("cat:")
print(torch.cat([a, b]))

print("\nstack:")
print(torch.stack([a, b]))
````

Compare their shapes:

```text
cat   → (4,)
stack → (2, 2)
```

`stack()` adds a new dimension, while `cat()` extends an existing dimension.

# 18. CPU and GPU

One of PyTorch's most important capabilities is hardware acceleration.

A tensor can live on:

```text
CPU
GPU
```

You can inspect the current device using:

```python
tensor.device
```

````python
x = torch.tensor([1, 2, 3])

print("Device:", x.device)
````

## Detecting CUDA

`torch.cuda.is_available()` tells us whether CUDA GPU support is available to PyTorch.

````python
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Running with CPU only.")
````

### Selecting a device

A common pattern is:

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

This allows the same code to work on both machines with and without a CUDA-capable GPU.

````python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Selected device:", device)
````

## Moving a tensor to the device

Use:

```python
tensor.to(device)
```

or:

```python
tensor.cuda()
```

`to(device)` is generally more flexible because it can work with either CPU or GPU.

````python
x = torch.tensor([1., 2., 3.])

x = x.to(device)

print(x)
print("Device:", x.device)
````

### Important rule

When performing operations between tensors, they generally need to be on compatible devices.

For example, a CPU tensor and CUDA tensor cannot normally be directly combined.

````python
cpu_tensor = torch.tensor([1., 2., 3.])

if torch.cuda.is_available():
    gpu_tensor = torch.tensor([4., 5., 6.], device="cuda")

    try:
        print(cpu_tensor + gpu_tensor)
    except RuntimeError as e:
        print("Expected device mismatch error:")
        print(e)
else:
    print("CUDA is not available; device mismatch experiment skipped.")
````

# 19. Tensor Conversion

## Tensor → NumPy

For CPU tensors, `.numpy()` can convert a tensor to a NumPy array.

````python
x = torch.tensor([1, 2, 3])

numpy_array = x.numpy()

print("PyTorch tensor:", x)
print("NumPy array:", numpy_array)
print("NumPy type:", type(numpy_array))
````

## NumPy → Tensor

Use:

```python
torch.from_numpy(array)
```

````python
import numpy as np

array = np.array([10, 20, 30], dtype=np.float32)

tensor = torch.from_numpy(array)

print("NumPy:", array)
print("Tensor:", tensor)
print("Tensor dtype:", tensor.dtype)
````

### Important

`torch.from_numpy()` can share memory with the NumPy array.

That means modifying one may affect the other.

This is different from creating a completely independent copy.

````python
array = np.array([1, 2, 3], dtype=np.float32)
tensor = torch.from_numpy(array)

tensor[0] = 999

print("Tensor:", tensor)
print("NumPy array:", array)
````

# 20. Random Seeds and Reproducibility

Random initialization is common in machine learning.

If you want repeatable experiments, set a seed:

```python
torch.manual_seed(42)
```

The seed does not make values "better". It makes the random sequence reproducible.

````python
torch.manual_seed(42)

a = torch.rand(5)

torch.manual_seed(42)

b = torch.rand(5)

print("a:", a)
print("b:", b)
print("Same:", torch.equal(a, b))
````

Try changing `42` to another number.

The generated values will change, but running the same seed again will reproduce the same sequence under the same relevant environment conditions.

# 22. Common Tensor Errors

Understanding errors is part of learning PyTorch.

### Error 1 — Invalid reshape

The number of elements must remain unchanged.

### Error 2 — Device mismatch

CPU and CUDA tensors generally cannot be directly combined.

### Error 3 — dtype mismatch

Some operations require compatible numerical types.

### Error 4 — Incorrect dimensions

Many neural-network layers expect a specific input shape.

### Error 5 — Integer tensors where floating-point tensors are expected

Neural-network computations frequently require floating-point inputs.

# 23. Practical Example — Image Tensor

Let's represent one RGB image.

A common PyTorch image layout is:

```text
(C, H, W)
```

where:

- `C` = channels
- `H` = height
- `W` = width

For an RGB image:

```text
C = 3
```

A `224 × 224` RGB image therefore has:

```text
(3, 224, 224)
```

````python
image = torch.rand(3, 224, 224)

print("Image shape:", image.shape)
print("Number of dimensions:", image.ndim)
print("Number of values:", image.numel())
print("Data type:", image.dtype)
print("Device:", image.device)
````

## Add a batch dimension

Neural networks commonly process multiple images at once.

For example, 32 images:

```text
(batch, channels, height, width)
```

becomes:

```text
(32, 3, 224, 224)
```

````python
images = torch.rand(32, 3, 224, 224)

print("Batch shape:", images.shape)
print("Number of images:", images.shape[0])
print("Channels:", images.shape[1])
print("Height:", images.shape[2])
print("Width:", images.shape[3])
````
