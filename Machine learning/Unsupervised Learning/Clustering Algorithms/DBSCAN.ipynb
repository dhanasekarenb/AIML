# DBSCAN

```python
import numpy as np
```

# Data

```python
X = np.array([
    [1, 2],
    [2, 2],
    [2, 3],
    [8, 8],
    [8, 9],
    [25, 80]
])
```

# Parameter

for Labels
| Value | Meaning    |
| ----- | ---------- |
| -1    | Noise      |
| 0     | Unassigned |
| 1     | Cluster 1  |
| 2     | Cluster 2  |


```python
eps = 2 # How far should we search for neighbors?
min_samples = 2 # Minimum number of nearby points required to form a dense region.

visited = [False] * len(X)
labels = [0] * len(X)
cluster_id = 0
```

# Find index using point

```python
def pos(point):
    return X.index(point)
```

# Distance

```python
def euclidean_distance(x, point):
    return np.sqrt(np.sum((np.array(x) - np.array(point)) ** 2))

# euclidean_distance(X[0], (2,3))
```

# Region Query 

```python
def region_query(point_idx):
    neighbors = []

    for i in range(len(X)):
        if euclidean_distance(X[point_idx], X[i]) <= eps:
            neighbors.append(i)

    return neighbors
```

```python
def expand_cluster(point_idx, neighbors, cluster_id):


    # set label value for point
    labels[point_idx] = cluster_id

    i = 0

    # iterate all neighbors
    while i < len(neighbors):
        
        # get neighbor index 
        neighbor_idx = neighbors[i]

        # check visited if False
        if not visited[neighbor_idx]:

            # Update visisted for current neighbor
            visited[neighbor_idx] = True

            # find sub neigbors for that neighbor
            sub_neighbors = region_query(neighbor_idx)

            # check cluster
            if len(sub_neighbors) >= min_samples:
                
                # checking whether the sub-neigbors are in neigbors if not add them in cluster
                for n in sub_neighbors:
                    if n not in neighbors:
                        neighbors.append(n)

        if labels[neighbor_idx] <= 0:
            labels[neighbor_idx] = cluster_id

        i += 1
```

```python
def dbscan():

    # Initial cluster id 0    
    cluster_id = 0

    # iterate all elements from X using index
    for point_idx in range(len(X)):

        # check for whether visited the point or not
        if visited[point_idx]:
            continue

        # if not visited mark as visited and perform
        visited[point_idx] = True

        # find all neighbors for the point
        neighbors = region_query(point_idx)

        # if not cluster
        if len(neighbors) < min_samples:

            labels[point_idx] = -1

        # if cluster
        else:

            cluster_id += 1

            expand_cluster(
                point_idx,
                neighbors,
                cluster_id
            )

    return labels
```

```python
result = dbscan()

print("Labels:", result)
```
