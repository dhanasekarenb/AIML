## Why Seaborn Exists

Matplotlib is powerful but requires more code.

Seaborn provides:

- Better defaults
- Statistical visualizations
- DataFrame integration
- Faster EDA

AI engineers use Seaborn heavily before model training.


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Seaborn vs Matplotlib

### Matplotlib
- Low level
- More control

### Seaborn
- High level
- Faster EDA
- Better looking plots

## Themes

Available themes:

- darkgrid
- whitegrid
- dark
- white
- ticks


```python
sns.set_theme(style="ticks")

plt.plot([1,2,3],[2,4,6])
plt.title("Ticks Theme")
plt.show()
```

## Loading Built-in Datasets

Seaborn includes datasets for learning and practice.


```python
tips = sns.load_dataset("tips")

tips.head()
```

```python
tips.info()
```

```python
tips.describe()
```

## Axes-Level Functions

Operate on a specific matplotlib axis.

Examples:

- scatterplot
- lineplot
- boxplot
- histplot


```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
```

## Figure-Level Functions

Create and manage their own figure.

Examples:

- relplot (relationship plot at figure level)
- displot (distribution plot at figure level)
- catplot (categorical plot at figure level)
- pairplot (multi-feature figure-level plot)


```python
sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    col = "sex"
)
```

```python
sns.displot(
    data = tips,
    x = "total_bill",
    y = "tip",
    col = "sex"
)
```

```python
sns.catplot(
    data = tips,
    x = "day",
    y = "total_bill",
    kind = "bar",
    col = "sex",
    # hue = "smoker"
)
```

```python
sns.pairplot(
    data = tips,
    hue = "smoker"
)
```

## DataFrame Integration

Seaborn works directly with Pandas DataFrames.


```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Total Bill vs Tip")
plt.show()
```
