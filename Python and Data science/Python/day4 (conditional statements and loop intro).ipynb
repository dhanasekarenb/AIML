# Else

```python
# we can write a if without else
# but without if we cant write a else
# we can use if, elif, elif ... n without else

# example (wrong)
# else: 
#     print("nothing")

# then it's a default condition

# syntax

# if condition:
    # statement 1
    # statement n
# else: (default)
    # statement 1
    # statement n


# program

# if cost >= totalcost then discount will apply
discount = 50
total_cost = 10000

actual_cost = float(input("Enter the total cost : "))
if total_cost <= actual_cost:
    print("discount will apply 50%")
    print(f"total cost {actual_cost * (discount/100)}")

else:
    print("sorry better luck next time")
```

# Elif

```python
# elif is nothing but else if


# example
# problem 1
# check cost > 500000 print rich or cost >200000 middle class or cost < 100000 poor

cost = 250000

if cost > 500000:
    print("rich")
else:
    if cost > 200000:
        print("middle class")
    else:
        if cost < 100000:
            print("poor")

# Elif

if cost > 500000:
    print("rich")

elif cost > 200000:
    print("middle class")

elif cost < 100000:
    print("poor")
```

# Nested if

```python
# syntax 

# if condition:
    # statement 1
    # if condition:
        # statement 1
        # if condition:
            # statement 1
        # statement 2
    # statement 2
    # statement n




# problem 
# student marks based scholarship

marks = 60
school = "gov"
clg_bus = True

if marks >= 60:
    if school == "gov":
        if clg_bus == True:
            print("ok for scholarship  but need to pay 5000 extra")
        else:
            print("no need for extra pay scholarship applied")
    else:
        print("need to finish your 12 in gov school")
else:
    print("no scholarship")
```

```python
# for

# we know when to stop

# for example 
# counting from 0 to 100

# syntax
# for i in range(0, 10, 1):
    # statement

# for -> keyword
# i -> var
# in -> keyword
# range -> function
    # 0 -> start
    # 10 -> end
    # 1 -> step

# simple program
for i in range(1, 10):
    print(i)
```

```python
for i in range(10):
    print(i)
```

```python
for j in range(5, 15, 2):
    print(j)
```

```python
for rev in range(20, 10, -1):
    print(rev)
```

```python
# we can iterate a string or seconday datatype

name = "yaazhini"

for n in name:
    print(n)
```

```python
# list
l = ["xyz", "1", "2"]

for x in l:
    print(x)
```

```python
n = 0

while True:
    print("1")
    n = n + 1

    if n > 10:
        print("come out of loop")
        break
```

```python
while True:
    flag = input("write True to stop ")

    if flag == "True":
        break
    else:
        print("thanks for not stoping ")
```

```python
flag = True

while flag:
    x = input("write True to stop ")

    if x == "True":
        flag = False
    else:
        print("thanks for not stoping ")
```
