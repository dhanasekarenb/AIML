# What is a Subplot?

A subplot is simply multiple axes inside a single figure.

```python
fig, ax = plt.subplots()
```

creates one subplot.

```python
fig, ax = plt.subplots(2,2)
```

creates four subplots.


````python

import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot([1,2,3],[4,5,6])

plt.show()
````


# Creating Multiple Subplots

Example:

2 rows × 2 columns


````python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    2,
    2,
    figsize=(8,6)
)

plt.show()
````


# Accessing Individual Axes

Each subplot has its own axis object.


````python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2)

ax[0,0].plot([1,2,3],[1,4,9])
ax[0,1].plot([1,2,3],[1,2,3])
ax[1,0].plot([1,2,3],[3,2,1])
ax[1,1].plot([1,2,3],[9,4,1])

plt.show()
````


# AI Example — Training Dashboard

A common deep learning dashboard contains:

- Loss Curve
- Accuracy Curve


````python

import matplotlib.pyplot as plt

epochs = [1,2,3,4,5]

loss = [2.5,1.8,1.2,0.8,0.5]
accuracy = [55,65,75,85,92]

fig, ax = plt.subplots(
    1,
    2,
    figsize=(10,4)
)

ax[0].plot(epochs, loss)
ax[0].set_title("Loss")

ax[1].plot(epochs, accuracy)
ax[1].set_title("Accuracy")

plt.show()
````


# Training vs Validation Dashboard

One of the most common AI visualizations.


````python

import matplotlib.pyplot as plt

epochs = [1,2,3,4,5]

train_loss = [2.5,1.8,1.2,0.8,0.5]
val_loss = [2.6,2.0,1.5,1.4,1.7]

train_acc = [55,65,75,85,92]
val_acc = [54,63,72,76,74]

fig, ax = plt.subplots(
    2,
    1,
    figsize=(8,8)
)

ax[0].plot(epochs, train_loss, label="Train")
ax[0].plot(epochs, val_loss, label="Validation")
ax[0].legend()
ax[0].set_title("Loss")

ax[1].plot(epochs, train_acc, label="Train")
ax[1].plot(epochs, val_acc, label="Validation")
ax[1].legend()
ax[1].set_title("Accuracy")

plt.show()
````


# Histograms Inside Dashboards

EDA often requires multiple distributions.


````python

import numpy as np
import matplotlib.pyplot as plt

age = np.random.normal(30, 5, 1000)
salary = np.random.normal(50000, 10000, 1000)

fig, ax = plt.subplots(
    1,
    2,
    figsize=(10,4)
)

ax[0].hist(age)
ax[0].set_title("Age")

ax[1].hist(salary)
ax[1].set_title("Salary")

plt.show()
````


# Comparing Multiple Models

Subplots make comparisons easy.


````python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    1,
    3,
    figsize=(12,4)
)

ax[0].bar(["A"], [82])
ax[0].set_title("Model A")

ax[1].bar(["B"], [89])
ax[1].set_title("Model B")

ax[2].bar(["C"], [93])
ax[2].set_title("Model C")

plt.show()
````


# Tight Layout

Problem:

Labels overlap.

Solution:

```python
plt.tight_layout()
```


````python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2)

plt.tight_layout()

plt.show()
````


# Figure-Level Title

Useful for dashboards.


````python

import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2)

fig.suptitle(
    "AI Dashboard"
)

plt.show()
````


# Professional AI Dashboard

This resembles what you may see in:

- Kaggle notebooks
- ML reports
- Research papers


````python

import matplotlib.pyplot as plt

epochs = [1,2,3,4,5]

loss = [2.5,1.8,1.2,0.8,0.5]
accuracy = [55,65,75,85,92]

fig, ax = plt.subplots(
    2,
    2,
    figsize=(10,8)
)

ax[0,0].plot(epochs, loss)
ax[0,0].set_title("Loss")

ax[0,1].plot(epochs, accuracy)
ax[0,1].set_title("Accuracy")

ax[1,0].bar(["A","B","C"], [82,89,93])
ax[1,0].set_title("Model Comparison")

ax[1,1].hist(loss)
ax[1,1].set_title("Loss Distribution")

fig.suptitle("Training Dashboard")

plt.tight_layout()

plt.show()
````
