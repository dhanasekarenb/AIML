# What is a Histogram?

A histogram groups values into ranges called bins.

Instead of plotting individual values, it shows:

- How many values fall into each range
- Shape of the distribution

### AI Usage

Before training:
- Check feature distributions
- Identify skewness
- Detect abnormal values


````python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=30)

plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
````


# Understanding Bins

Bins divide the data into intervals.

Example:

```text
0-10
10-20
20-30
30-40
```

The histogram counts how many samples fall inside each interval.

More bins → more detail

Fewer bins → smoother overview


````python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=10)
plt.title("10 Bins")
plt.show()
````

````python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=50)
plt.title("50 Bins")
plt.show()
````


# Normal Distribution

The most famous distribution in statistics.

Characteristics:

- Bell-shaped curve
- Mean near center
- Symmetrical

Many machine learning assumptions originate from normal distributions.


````python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(
    loc=0,
    scale=1,
    size=10000
)

plt.hist(data, bins=50)

plt.title("Normal Distribution")

plt.show()
````


# Why AI Engineers Care About Normal Distributions

Algorithms often perform better when data is approximately normal.

Examples:

- Linear Regression
- Logistic Regression
- Naive Bayes
- PCA

Not mandatory, but useful.



# Right Skewed Distribution

Most values are small.

Few large values exist.

Common examples:

- Salaries
- House prices
- Income


````python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.exponential(
    scale=1,
    size=5000
)

plt.hist(data, bins=50)

plt.title("Right Skewed Distribution")

plt.show()
````


# Left Skewed Distribution

Most values are large.

Few small values exist.


````python

import numpy as np
import matplotlib.pyplot as plt

data = -np.random.exponential(
    scale=1,
    size=5000
)

plt.hist(data, bins=50)

plt.title("Left Skewed Distribution")

plt.show()
````


# Uniform Distribution

Every value appears with similar probability.


````python

import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(
    low=0,
    high=10,
    size=5000
)

plt.hist(data, bins=50)

plt.title("Uniform Distribution")

plt.show()
````


# Outlier Detection

Outliers can negatively affect models.

Example:

```python
[10, 11, 12, 13, 1000]
```

Histogram makes unusual values visible.


````python

import matplotlib.pyplot as plt

data = [10,11,12,13,14,15,1000]

plt.hist(data)

plt.title("Outlier Example")

plt.show()
````


# Normalization

Used when feature scales differ greatly.

Example:

```python
Age     = 20-80
Salary  = 20000-100000
```

Without normalization:

Salary dominates.

After normalization:

Both contribute fairly.



# Standardization

Formula:



\(
z = \frac{x - \mu}{\sigma}
\)

Purpose:

- Mean becomes 0
- Standard deviation becomes 1

Used heavily in Machine Learning.
