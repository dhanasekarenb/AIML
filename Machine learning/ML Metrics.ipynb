#### 1. Accuracy
$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

When to use:

When classes are balanced

When all mistakes cost the same

Not good for: Imbalanced datasets (e.g., fraud detection)

#### 2. Precision

$$
Precision = \frac{TP}{TP + FP}
$$

Meaning: Out of items predicted positive, how many were actually positive?

When to use:

When false positives are bad

Examples:

✔ Spam detection

✔ Disease diagnosis (don’t say someone is sick unless sure)

#### 3. Recall
$$
Recall = \frac{TP}{TP + FN}
$$
Meaning: Out of real positives, how many did we catch?

When to use:

When false negatives are dangerous

Examples:

✔ Cancer detection

✔ Security alert systems

#### 4. F1 score
$$
F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}
$$

Meaning: Harmonic mean of Precision + Recall.

When to use:

Imbalanced data

When both FP and FN matter equally

# 2. Object Detection Metrics

#### 1. IoU — Intersection over Union
$$
IoU = \frac{Area\ of\ Overlap}{Area\ of\ Union}
$$

Meaning: How much predicted box overlaps with ground truth box.

Use When:

Evaluating bounding box quality

Used in object detection pipelines (YOLO, Faster RCNN)

image.png

#### 2. mAP — mean Average Precision

mAP = Average precision over all classes.

$$
mAP = \frac{1}{N} \sum_{i=1}^{N} AP_i
$$

Meaning:

Measures BOTH precision and recall across all confidence thresholds

Use When:

Evaluating object detection models

Standard in COCO, Pascal VOC datasets

# 3. Segmentation Metrics

#### 1. mIoU — Mean Intersection over Union
$$
IoU = \frac{TP}{TP + FP + FN}
$$
@
$$
mIoU = mean(IoU\ across\ all\ classes)
$$
Meaning: Pixel-level overlap between predicted mask and ground truth.

Use When:

Semantic segmentation

Image segmentation tasks (UNet, DeepLab)

image.png

#### . Dice Coefficient / Dice Loss

##### Dice Coefficient
$$
Dice = \frac{2TP}{2TP + FP + FN}
$$

##### Dice Loss
$$
Loss_{Dice} = 1 - Dice
$$

Meaning: Similar to IoU but more sensitive to small objects.

Use When:

Medical imaging (tumor segmentation)

Small or imbalanced segmentation masks

# 4. Regression Metrics


#### 1. Mean Squared Error (MSE)

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y_i})^2
$$

Meaning: Punishes big errors more heavily.

Use When:

You want stronger penalty for large mistakes

Linear Regression, forecasting

#### 2. Mean Absolute Error (MAE)

$$
MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y_i}|
$$
Meaning: Measures average absolute error.

Use When:

Outliers exist (robust metric)

You want equal weight for all errors


#### 3. Root Mean Squared Error (RMSE)

$$
RMSE = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y_i})^2 }
$$
