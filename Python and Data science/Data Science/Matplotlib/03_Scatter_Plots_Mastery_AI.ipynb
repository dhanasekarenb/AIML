# Module 3 — Scatter Plots Mastery for AI & Data Analytics

Scatter plots are one of the most important visualizations in Machine Learning.

## Learning Goals

- Understand relationships between variables
- Detect correlations
- Identify clusters
- Find outliers
- Build intuition for ML algorithms
- Visualize datasets before training models


## What is a Scatter Plot?

A scatter plot displays individual data points on a coordinate plane.

Each point represents:

```text
(x, y)
```

### Why do we use it?

Because we want to understand how one feature affects another.


````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,5,4,5]

plt.scatter(x, y)
plt.show()
````

## Why AI Engineers Use Scatter Plots

Before training a model we ask:

- Are two features related?
- Is there a pattern?
- Are there outliers?
- Can classes be separated?

Scatter plots answer these questions quickly.


## Positive Correlation

As X increases, Y increases.


````python
import matplotlib.pyplot as plt

age = [20,25,30,35,40]
salary = [25000,35000,45000,55000,65000]

plt.scatter(age, salary)

plt.xlabel("Age")
plt.ylabel("Salary")

plt.title("Positive Correlation")

plt.show()
````

### Interpretation

Older employees tend to have higher salaries.

This is a positive relationship.


## Negative Correlation

As X increases, Y decreases.


````python
hours_tv = [1,2,3,4,5,6]
exam_score = [95,90,85,80,75,70]

plt.scatter(hours_tv, exam_score)

plt.xlabel("TV Hours")
plt.ylabel("Exam Score")

plt.title("Negative Correlation")

plt.show()
````

## No Correlation

Sometimes variables are unrelated.


````python
import numpy as np
import matplotlib.pyplot as plt

x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x, y)
plt.title("No Correlation")

plt.show()
````

## Outlier Detection

Outliers can damage model performance.

Scatter plots make them easy to spot.


````python
x = [1,2,3,4,5,6,7,8]
y = [10,12,15,17,18,20,22,100]

plt.scatter(x,y)

plt.title("Outlier Detection")

plt.show()
````

## Clustering Intuition

Scatter plots help understand clustering algorithms such as:

- K-Means
- DBSCAN
- Hierarchical Clustering


````python
import matplotlib.pyplot as plt

x = [1,2,1.5,8,9,8.5]
y = [1,2,1.8,8,9,8.2]

plt.scatter(x,y)

plt.title("Cluster Visualization")

plt.show()
````

## Understanding KNN Using Scatter Plots

KNN classifies based on nearby points.

Scatter plots make neighborhood relationships visible.


````python
import matplotlib.pyplot as plt

class_a_x = [1,2,2,3]
class_a_y = [1,1,2,2]

class_b_x = [7,8,8,9]
class_b_y = [7,7,8,8]

plt.scatter(class_a_x, class_a_y, label="Class A")
plt.scatter(class_b_x, class_b_y, label="Class B")

plt.legend()
plt.show()
````

## Class Separation

One of the most important tasks in Machine Learning.

Can a model separate classes?

Scatter plots help answer this.


````python
plt.scatter([1,2,3],[1,2,1], label="Class 0")
plt.scatter([7,8,9],[7,8,7], label="Class 1")

plt.legend()
plt.title("Class Separation")

plt.show()
````

## Feature Engineering

Scatter plots help discover:

- Linear relationships
- Non-linear relationships
- Transformations needed


````python
import numpy as np

x = np.arange(0,10,0.5)
y = x**2

plt.scatter(x,y)

plt.title("Non Linear Relationship")

plt.show()
````

## Professional Scatter Plot Formatting

````python
import matplotlib.pyplot as plt

age = [20,25,30,35,40]
salary = [25000,35000,45000,55000,65000]

plt.figure(figsize=(8,5))

plt.scatter(
    age,
    salary,
)

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")

plt.grid(True)

plt.show()
````
