```python
sample = "sAmPle text"


print(sample.lower())

print(sample.upper())

print(sample.capitalize())

print(sample.title())
```

```python
data = "     hello     "


print(data.strip())

print(data.lstrip())

print(data.rstrip())
```

```python
data = "Good Morning did you have you're coffee."

l = []
temp = ""
for i in data:
    if i == " " or i == ".":
        l.append(temp)
        temp = ""
    else:
        temp += i

print(l)
```

```python
data = "Good Morning did you have you're coffee"
data_2 = "test1,test2,test3,test4"

print(data.split())
print(data_2.split(','))
```

```python
data = "WBT001"
# data = "WBT"
print(data.isalpha())#check for only alphabets
print(data.isdigit()) #check for only digits
print(data.isalnum()) #check for alpa and digits if one exit it will return t
print(data.isspace()) #check for space
```

```python
if data.isalpha() == False and data.isalnum() == True:
    print("correct")
```
