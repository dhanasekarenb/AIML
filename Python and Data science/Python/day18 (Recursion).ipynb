# Recursion

Calling a function inside a function

### Syntax

```python
n = 1

def func_name(n):

    if (n > 10):
        return n

    # statement 1 to n
    print(n)
    func_name(n+1)

func_name(n)
```

```python
def test(n):
    
    if n > 5:
        print(f"return {n}")
        return 0
    
    print(n)

    test(n+1)

    print(f"this is not the end {n}")

test(1)
```
