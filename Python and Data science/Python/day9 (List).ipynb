```python
a = [5, 6, 3, 8, 2, 1, 7, 1]
b = [8, 2, 1] # sublist
a_str = ','.join(map(str, a)) # Convert to string representation
b_str = ','.join(map(str, b))
res = a_str.find(b_str) != -1

print(a_str)
print(b_str)

print(res)
```
