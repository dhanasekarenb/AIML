# Displaying an Image

The most basic operation.


````python

import matplotlib.pyplot as plt
import numpy as np

image = np.random.rand(224,224,3)

plt.imshow(image)
plt.title("Image")

plt.show()
````


# Understanding Image Shape

Typical image:

```python
(H, W, C)
```

Example:

```python
(224, 224, 3)
```

Where:

- H = Height
- W = Width
- C = Channels


````python

import numpy as np

image = np.random.rand(
    224,
    224,
    3
)

print(image.shape)
````


# RGB Channels

Most images contain:

- Red
- Green
- Blue

Each channel stores intensity values.


````python

import matplotlib.pyplot as plt
import numpy as np

image = np.random.rand(224,224,3)

plt.imshow(image[:,:,0])
plt.title("Red Channel")

plt.show()
````


# Visualizing All Channels

Useful for understanding image data.


````python

import matplotlib.pyplot as plt
import numpy as np

image = np.random.rand(224,224,3)

fig, ax = plt.subplots(1,3, figsize=(10,4))

ax[0].imshow(image[:,:,0])
ax[0].set_title("Red")

ax[1].imshow(image[:,:,1])
ax[1].set_title("Green")

ax[2].imshow(image[:,:,2])
ax[2].set_title("Blue")

plt.show()
````


# Grayscale Images

Shape:


Common in:

- Classical CV


````python

import numpy as np
import matplotlib.pyplot as plt

gray = np.random.rand(224,224)

plt.imshow(gray, cmap="gray")

plt.title("Grayscale")

plt.show()
````


# Comparing Original and Processed Images

Common during preprocessing.


````python

import numpy as np
import matplotlib.pyplot as plt

original = np.random.rand(224,224,3)
processed = np.clip(original * 0.5, 0, 1)

fig, ax = plt.subplots(1,2, figsize=(8,4))

ax[0].imshow(original)
ax[0].set_title("Original")

ax[1].imshow(processed)
ax[1].set_title("Processed")

plt.show()
````


# Understanding Image Tensors in PyTorch

PyTorch uses:

```python
(C, H, W)
```

instead of:

```python
(H, W, C)
```

This causes confusion for beginners.


````python

import torch

image = torch.rand(
    3,
    224,
    224
)

print(image.shape)
````


# Visualizing PyTorch Tensors

Need to convert:

```python
(C,H,W)

→

(H,W,C)
```


````python

import torch
import matplotlib.pyplot as plt

image = torch.rand(
    3,
    224,
    224
)

image = image.permute(
    1,
    2,
    0
)

plt.imshow(image)

plt.show()
````


# Object Detection Visualization

Object detection predicts bounding boxes.

Examples:

- YOLO
- Faster R-CNN
- DETR


````python

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Rectangle

image = np.random.rand(224,224,3)

fig, ax = plt.subplots()

ax.imshow(image)

rect = Rectangle(
    (50,50),
    80,
    100,
    fill=False
)

ax.add_patch(rect)

plt.show()
````


# Multiple Bounding Boxes

Real object detection contains many boxes.


````python

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Rectangle

image = np.random.rand(224,224,3)

fig, ax = plt.subplots()

ax.imshow(image)

boxes = [
    (30,30,60,80),
    (120,40,50,90),
]

for x,y,w,h in boxes:
    ax.add_patch(
        Rectangle(
            (x,y),
            w,
            h,
            fill=False
        )
    )

plt.show()
````


# Segmentation Masks
