# KNN (K-Nearest Neighbour)

```python
import numpy as np
```

# Data

```python
X = [
    [1, 2],
    [8, 8],
    [9, 8],
    [8, 9],
    [1, 1],
    [2, 1]
]

y = [
    "A",
    "B",
    "B",
    "B",
    "A",
    "A"
]
```

# Parameter

```python
k = 3
```

# KNN Algorithm

```python
class KNN:
    
    def __init__(self, k):
        self.k = k


    def fit(self, X, y):
        
        self.X = X
        self.y = y

    def predict(self, x):
        
        distance = []

        for i in range(len(self.X)):
            distance.append(self.euclidean_distance(self.X[i], x))

        # argsort sort the value and return the sorted index
        distance = np.array(distance)
        sorted_index = np.argsort(distance)
        print(sorted_index)

        vote = []

        for i in range(self.k):
            vote.append(self.y[sorted_index[i]])
        
        # vote.append('B')
        # print(vote)
        print(max(set(vote), key=vote.count))

    def euclidean_distance(self, x, point):
        
        return np.sqrt(np.sum((np.array(x) - np.array(point)) ** 2))
```

```python
knn = KNN(k)

knn.fit(X, y)

knn.predict([2, 2])

# knn.euclidean_distance([3,3], [1, 1])
```
