# What is a Bar Chart?

A bar chart represents values using rectangular bars.

### AI Usage

- Model comparison
- Feature importance
- Dataset comparison
- Experiment tracking
- Hyperparameter studies


````python

import matplotlib.pyplot as plt

models = ["LR", "RF", "XGB"]
accuracy = [82, 89, 93]

plt.bar(models, accuracy)

plt.title("Model Accuracy Comparison")

plt.show()
````


# Understanding X and Y

# Vertical Bar Chart

Default bar chart.


````python

import matplotlib.pyplot as plt

products = ["A", "B", "C", "D"]
sales = [100, 150, 120, 180]

plt.bar(products, sales)

plt.title("Product Sales")

plt.show()
````


# Horizontal Bar Chart

Useful when labels are long.


````python

import matplotlib.pyplot as plt

features = [
    "Age",
    "Income",
    "Experience",
    "Education"
]

importance = [0.25, 0.40, 0.20, 0.15]

plt.barh(features, importance)

plt.title("Feature Importance")

plt.show()
````


# Feature Importance

One of the most common AI applications.

Questions:

Which feature contributes most?

Which feature contributes least?


````python

import matplotlib.pyplot as plt

features = ["Age","Salary","Experience","Education"]
importance = [0.15,0.50,0.25,0.10]

plt.bar(features, importance)

plt.xlabel("Features")
plt.ylabel("Importance")

plt.title("Feature Importance")

plt.show()
````


# Comparing Machine Learning Models

Very common during experimentation.


````python

models = [
    "Logistic Regression",
    "Random Forest",
    "XGBoost"
]

accuracy = [85, 91, 94]

plt.figure(figsize=(8,5))

plt.bar(models, accuracy)

plt.ylabel("Accuracy (%)")
plt.title("Model Comparison")

plt.show()
````


# Grouped Bar Charts

Used when comparing multiple metrics.


````python

import numpy as np
import matplotlib.pyplot as plt

models = ["RF", "XGB", "SVM"]

accuracy = [90,94,88]
precision = [88,93,86]

x = np.arange(len(models))
width = 0.35

plt.bar(
    x - width/2,
    accuracy,
    width,
    label="Accuracy"
)

plt.bar(
    x + width/2,
    precision,
    width,
    label="Precision"
)

plt.xticks(x, models)

plt.legend()

plt.show()
````


# Experiment Tracking

Imagine testing:

- Learning Rate 0.1
- Learning Rate 0.01
- Learning Rate 0.001

Bar charts compare results quickly.


````python

learning_rates = [
    "0.1",
    "0.01",
    "0.001"
]

accuracy = [82,91,87]

plt.bar(
    learning_rates,
    accuracy
)

plt.title("Learning Rate Comparison")

plt.show()
````


# Adding Colors

Colors improve readability.


````python

models = ["LR","RF","XGB"]
accuracy = [82,89,93]

plt.bar(
    models,
    accuracy,
    color=["red","blue","green"]
)

plt.show()
````


# Adding Value Labels

Useful in presentations and reports.


````python

models = ["LR","RF","XGB"]
accuracy = [82,89,93]

bars = plt.bar(models, accuracy)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        str(bar.get_height()),
        ha='center'
    )

plt.show()
````


# Professional Formatting


````python

models = ["LR","RF","XGB"]
accuracy = [82,89,93]

plt.figure(figsize=(8,5))

plt.bar(models, accuracy)

plt.xlabel("Models")
plt.ylabel("Accuracy (%)")
plt.title("Model Accuracy Comparison")

plt.grid(axis="y")

plt.show()
````
