# Network

```python
class QNetwork(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer_1 = nn.Linear(in_features = 2, out_features = 64)
        self.layer_2 = nn.Linear(in_features = 64, out_features = 4)

        self.relu = nn.ReLU()


    def forward(self, x):

        x = self.layer_1(x)

        x = self.relu(x)

        x = self.layer_2(x)

        return x
```

# Take Action

```python
state = torch.tensor(start)

def choose_action(state:tuple) -> int:
    
    model = QNetwork()

    output = model(state)

    action = torch.max(output)
    
    return action

choose_action(state)
```
