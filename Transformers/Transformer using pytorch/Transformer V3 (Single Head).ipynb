# Scaled Dot Product Attention
# Actual Class implementation

```python
class SelfAttention(nn.Module):
    
    # constructor
    def __init__(self, d_model):
        
        super().__init__()

        self.Wq = nn.Linear(in_features = d_model, out_features = d_model)
        self.Wk = nn.Linear(in_features = d_model, out_features = d_model)
        self.Wv = nn.Linear(in_features = d_model, out_features = d_model)

        self.d_model = d_model

    
    def forward(self, x):

        print("Input:", x.shape)

        Q = self.Wq(x)
        K = self.Wk(x)
        V = self.Wv(x)

        print("Q:", Q.shape)
        print("K:", K.shape)
        print("V:", V.shape)

        scores = (Q @ K.T) / math.sqrt(self.d_model)

        print("Scores:", scores.shape)

        weights = torch.softmax(scores, dim=-1)

        print("Weights:", weights.shape)

        output = weights @ V

        print("Output:", output.shape)

        return output

    # def forward(self, x):

    #     Q = self.Wq(x)
    #     K = self.Wk(x)
    #     V = self.Wv(x)

    #     # calc scores @ -> dotproduct
    #     scores = (Q @ K.T) / math.sqrt(self.d_model)

    #     # sigmoid
    #     weights = torch.softmax(scores, dim = -1) 

    #     # calc weights with V
    #     output = weights @ V

    #     return output
```

```python
tokens = torch.tensor([0, 1, 2])

embedding = nn.Embedding(
    num_embeddings = 3,
    embedding_dim = 8
)

x = embedding(tokens)

print(x.shape)
```

```python
attentation = SelfAttention(d_model = 8)

output = attentation(x)

print(output.shape)
```
