# GMM (Gaussian Mixture Model)

```python
import numpy as np

X = np.array([
    # Cluster-like group 1
    [1.0, 1.5],
    [1.5, 2.0],
    [2.0, 1.0],
    [2.5, 1.8],
    [1.8, 2.5],
    [2.2, 2.2],
    [1.2, 2.3],
    [2.8, 1.2],
    [1.7, 1.3],
    [2.3, 1.6],

    # Cluster-like group 2
    [7.0, 7.5],
    [7.5, 8.0],
    [8.0, 7.0],
    [8.5, 7.8],
    [7.8, 8.5],
    [8.2, 8.2],
    [7.2, 8.3],
    [8.8, 7.2],
    [7.7, 7.3],
    [8.3, 7.6],

    # Cluster-like group 3
    [1.0, 8.0],
    [1.5, 8.5],
    [2.0, 7.5],
    [2.5, 8.2],
    [1.8, 8.8],
    [2.2, 8.3],
    [1.2, 8.7],
    [2.8, 7.7],
    [1.7, 7.8],
    [2.3, 8.1]
])

print(X.shape)
```

# Parameters

```python
K = 3               # Number of Clusters
N, D = X.shape      # N = samples and D = Features
```

# Initialize Means

```python
np.random.seed(42)

indices = np.random.choice(N, K, replace = False)

mu = X[indices].copy()
```

# Initialize Covariance Matrices

```python
covariances = np.array([np.eye(D) for _ in range(K)])

covariances
```

# Initialize Mixing Coefficients

```python
pi = np.ones(K) / K

pi
```

```python
# ==========================================
# 6. Print Initial Parameters
# ==========================================

print("Dataset shape:", X.shape)

print("\nInitial Means:")
print(mu)

print("\nInitial Covariances:")
print(covariances)

print("\nInitial Mixing Coefficients:")
print(pi)

print("\nSum of mixing coefficients:")
print(pi.sum())
```
