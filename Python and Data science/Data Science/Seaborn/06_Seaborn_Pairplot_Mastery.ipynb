# Module 6 — Pairplot Mastery

Pairplot is one of the most powerful Exploratory Data Analysis (EDA) tools in Seaborn.

A single pairplot can reveal:

- Correlations
- Relationships
- Clusters
- Outliers
- Class Separation

before any machine learning model is trained.

## Learning Goals

- Understand pairplot
- Analyze multiple features together
- Discover relationships automatically
- Visualize class separation
- Identify clusters
- Perform fast EDA


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
```

## Why Pairplot Matters

Imagine a dataset with:

- Age
- Salary
- Experience
- Education Score

Instead of creating many scatterplots manually,

Pairplot automatically creates all combinations.


## What is Pairplot?

Pairplot generates:

Feature A vs Feature B

Feature A vs Feature C

Feature B vs Feature C

and more.

It also displays distributions along the diagonal.


```python
iris_raw = load_iris(as_frame=True)

iris = iris_raw.frame.copy()

iris.columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "target"
]

iris["species"] = iris["target"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

iris = iris.drop(columns="target")

sns.pairplot(iris)
plt.show()
```

## Understanding the Output

Off-Diagonal:

Scatterplots

Diagonal:

Distributions

This allows us to study every feature pair simultaneously.


## Why AI Engineers Use Pairplots

Before model training:

- Detect correlations
- Detect separable classes
- Discover clusters
- Find unusual observations


## Pairplot with Hue

One of the most powerful EDA techniques.

Hue colors observations by class.


```python
sns.pairplot(
    iris, hue="species"
)

plt.show()
```

## Class Separation

Questions:

## Selecting Specific Features

Sometimes we only want a subset of features.


```python
sns.pairplot(
    iris,
    vars=[
        "sepal_length",
        "petal_length",
        "petal_width"
    ]
)

plt.show()
```

## Different Diagonal Visualizations

Default:

Histogram

Alternative:

KDE


```python
sns.pairplot(
    iris,
    diag_kind="kde"
)

plt.show()
```

## Pairplot for Outlier Detection

Outliers often appear as isolated points.

These may indicate:

- Errors
- Rare events
- Valuable information


```python
data = pd.DataFrame({
    "A":[1,2,3,4,5,100],
    "B":[2,4,6,8,10,150],
    "C":[5,6,7,8,9,200]
})

sns.pairplot(data)

plt.show()
```

## Pairplot vs Scatterplot

Scatterplot:

One relationship

Pairplot:

All relationships

Pairplot is excellent for initial EDA.


## Limitations of Pairplot

Not ideal for:

- Hundreds of features
- Very large datasets

In such cases:

- Correlation heatmaps
- Feature selection

become important.
