# Example

```python
# Calc

x = int(input("Enter x : "))
y = int(input("Enter y : "))

op = int(input("Enter op(+(1), -(2)) : "))


if op == 1:
    print(x+y)
elif op == 2:
    print(x-y)
else:
    print("enter correct op number")
```
