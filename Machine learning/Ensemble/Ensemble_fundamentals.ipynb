# Ensemble Learning

Ensemble learning is a machine learning technique that combines prediction from multiple models to produce a final prediction that can be more accurate and robust then an individual model.

each individual model is called a base learner or weak learner, depending on the predictive strength

Example:
- model 1 -> spam
- model 2 -> not spam
- model 3 -> spam

using majority voting, the final prediction is spam

# Why Ensemble?

A single model may make mistake incorrect predicitions because of
- noise 
- limited training data 
- high variance
- unsutable model stucture

# Types of Ensemble

1. Bagging (Bootstrap Aggregating)
2. Boosting
3. Stacking (Stacked Generalization)

# Bagging (Bootstrap Aggregating)

Multiple models are trained independently on different bootstrap samples of the training data. Their predictions are combined using voting or averaging.

Main goal: reduce variance.

Example: Random Forest.

# Boosting

Models are trained sequentially, with each new learner attempting to improve the ensemble by addressing errors or reducing the loss.

Main goal: build a strong predictor from sequential learners.

Examples: AdaBoost, Gradient Boosting, XGBoost.

# Stacking (Stacked Generalization)

Predictions from several base models are used as inputs to another model called a meta-learner, which learns how to combine them.

Main goal: learn an effective combination of different models.

# Types of combine predictions

### Classification - Hard Voting

- Each model predicts a class label. The class receiving the most votes becomes the final prediction.

### Classification - Soft Voting

- Each model outputs class probabilities. We average those predictions and select the class with highest average.

### Regression - Averaging

- Each value predicts a numerical value. the final result will be the average of values.
