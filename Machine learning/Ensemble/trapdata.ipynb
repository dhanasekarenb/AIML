```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_rows = 3000

# Generate features
cgpa = np.random.normal(7.5, 1.0, n_rows).clip(5.0, 10.0)
backlogs = np.random.choice([0, 1, 2, 3], size=n_rows, p=[0.7, 0.15, 0.1, 0.05])
internships = np.random.choice([0, 1, 2, 3], size=n_rows, p=[0.4, 0.4, 0.15, 0.05])

# Trap 2: Sentinel values (-999) for missing internship durations
internship_duration = np.random.choice([3, 6, 12, -999], size=n_rows, p=[0.3, 0.3, 0.2, 0.2])

# Trap 3: Messy text categories in Specialization
specs = np.random.choice(["Computer Science", "CS ", "Comp. Sci.", "IT", "Electronics"], size=n_rows)

# Target variable creation
placed_prob = (cgpa / 10.0) * 0.5 + (internships * 0.1) - (backlogs * 0.15)
placed = (np.random.rand(n_rows) < placed_prob).astype(int)

df = pd.DataFrame({
    'CGPA': np.round(cgpa, 2),
    'Backlogs': backlogs,
    'Internships': internships,
    'Internship_Duration_Months': internship_duration,
    'Specialization': specs,
    'Placed': placed
})

# Trap 1: Data Leakage Column
df['Placement_Confirmation_ID'] = df['Placed'].apply(lambda x: f"CONFIRMED_PASS_{np.random.randint(1000,9999)}" if x==1 else "PENDING_REJECT")

df.to_csv("placement_train.csv", index=False)
print("Dataset generated successfully with 4 hidden traps!")
```
