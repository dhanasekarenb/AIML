## What is a Line Plot?

A line plot connects data points using lines.

### Why do we use it?

Because humans understand trends better visually than through raw numbers.

Example:

```python
loss = [2.5, 2.0, 1.6, 1.2, 0.8]
```

Looking at numbers is difficult.

A line plot immediately shows whether loss is decreasing.


````python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y)
plt.show()
````

## Understanding X and Y

In AI:

X-axis → Epochs

Y-axis → Metric

Examples:

- Loss
- Accuracy
- Precision
- Recall
- F1 Score


## AI Example 1 — Training Loss Curve

This is the most common plot in Deep Learning.


````python
import matplotlib.pyplot as plt

epochs = [1,2,3,4,5,6,7]
loss = [2.4,1.9,1.5,1.1,0.8,0.6,0.5]

plt.plot(epochs, loss)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")

plt.show()
````

### Interpretation

Loss is decreasing.

This means the model is learning.


## AI Example 2 — Accuracy Curve

Unlike loss, accuracy should increase.


````python
epochs = [1,2,3,4,5,6,7]
accuracy = [55,62,70,78,84,89,92]

plt.plot(epochs, accuracy)

plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.title("Training Accuracy")

plt.show()
````

## Adding Markers

Markers show exact data points.


````python
plt.plot(
    [1,2,3,4],
    [10,20,25,30],
    marker='o'
)

plt.show()
````

## Custom Line Styles

Useful when comparing multiple models.


````python
plt.plot([1,2,3],[4,5,6], linestyle='--')
plt.show()
````

## Comparing Multiple Models

One of the most common AI tasks.


````python
epochs = [1,2,3,4,5]

model_a = [2.4,1.8,1.2,0.9,0.7]
model_b = [2.5,2.0,1.6,1.3,1.1]

plt.plot(epochs, model_a, label="Model A")
plt.plot(epochs, model_b, label="Model B")

plt.legend()
plt.show()
````

## Training vs Validation Loss

This plot helps detect overfitting.


````python
epochs = [1,2,3,4,5,6]

train_loss = [2.5,1.8,1.2,0.8,0.5,0.3]
val_loss = [2.6,2.0,1.5,1.4,1.8,2.2]

plt.plot(epochs, train_loss, label="Training Loss")
plt.plot(epochs, val_loss, label="Validation Loss")

plt.legend()
plt.show()
````

### What is Overfitting?

Training loss keeps decreasing.

Validation loss starts increasing.

The model memorizes training data and fails to generalize.


## Underfitting Visualization

Both curves remain poor.


````python
epochs = [1,2,3,4,5]

train_loss = [3.0,2.8,2.6,2.4,2.2]
val_loss = [3.1,2.9,2.8,2.6,2.5]

plt.plot(epochs, train_loss, label="Training")
plt.plot(epochs, val_loss, label="Validation")

plt.legend()
plt.show()
````

## Professional Plot Formatting

````python
epochs = [1,2,3,4,5,6]
loss = [2.5,1.8,1.3,0.9,0.7,0.5]

plt.figure(figsize=(8,5))

plt.plot(
    epochs,
    loss,
    marker='o',
    linewidth=2,
    label='Loss'
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss Curve")

plt.grid(True)
plt.legend()

plt.show()
````
