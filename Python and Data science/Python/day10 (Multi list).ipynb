# How to get input for multi dimensional array

```python
school = []

for classes in range(0, 2):
    classes = []

    for section in range(0, 3):
        section = []

        for student in range(0, 1):
            student = []
            student.append(input("Enter name :"))
            student.append(input("Enter roll :"))

            section.append(student)

        classes.append(section)

    school.append(classes)

print(school)
```
