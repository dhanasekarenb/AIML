## JointPlot

JointPlot combines:

- Scatterplot
- Distribution of X
- Distribution of Y

into a single visualization.


```python
tips = sns.load_dataset("tips")

sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
```

## Why JointPlot is Useful

It helps us understand:

- Relationship between variables
- Distribution of X
- Distribution of Y

all at once.


## JointPlot with KDE

Useful for density analysis.


```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="kde"
)

plt.show()
```

## JointPlot with Regression

Combines relationship analysis with a regression trend.


```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="reg"
)

plt.show()
```

## FacetGrid

FacetGrid creates multiple plots based on category values.

Think of it as automatic subgroup analysis.


```python
tips = sns.load_dataset("tips")

g = sns.FacetGrid(
    tips,
    col="sex"
)

g.map(
    plt.hist,
    "total_bill"
)

plt.show()
```

## Why FacetGrid Matters

Questions:

- How do males behave?
- How do females behave?
- Are patterns different?

FacetGrid answers these visually.


## Multiple Facets

Analyze more than one category.


```python
g = sns.FacetGrid(
    tips,
    row="time",
    col="sex"
)

g.map(
    plt.hist,
    "total_bill"
)

plt.show()
```

## PairGrid

PairGrid is a customizable version of PairPlot.

PairPlot is easy.

PairGrid is powerful.


```python
from sklearn.datasets import load_iris

iris_raw = load_iris(as_frame=True)

iris = iris_raw.frame.copy()

iris.columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "target"
]

iris["species"] = iris["target"].map({
    0: "setosa",
    1: "versicolor",
    2: "virginica"
})

iris = iris.drop(columns="target")

g = sns.PairGrid(iris)

g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)

plt.show()
```

## Why PairGrid Matters

Custom control over:

- Diagonal plots
- Upper triangle
- Lower triangle

Useful in advanced EDA.


```python
iris = sns.load_dataset("iris")

g = sns.PairGrid(iris)

g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)

plt.show()
```

## Publication Quality Styling

Professional reports require:

- Consistent themes
- Readable fonts
- Proper spacing
- Clear labels


```python
sns.set_theme(
    style="whitegrid"
)

sns.set_palette("deep")

tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.title("Professional Plot")

plt.show()
```
