## Dataset Overview Dashboard

Start by understanding the dataset.


```python
tips = sns.load_dataset("tips")

print(tips.shape)
print(tips.columns)

tips.head()
```

## Dashboard 1 — Distribution Analysis

Questions:

- Is data normal?
- Is data skewed?
- Are there outliers?


```python
tips = sns.load_dataset("tips")

fig, ax = plt.subplots(1,2, figsize=(12,4))

sns.histplot(
    tips["total_bill"],
    kde=True,
    ax=ax[0]
)

ax[0].set_title("Bill Distribution")

sns.boxplot(
    x=tips["total_bill"],
    ax=ax[1]
)

ax[1].set_title("Bill Outliers")

plt.tight_layout()
plt.show()
```

## Dashboard 2 — Categorical Analysis

Questions:

- Which category is most common?
- Are classes balanced?


```python
fig, ax = plt.subplots(1,2, figsize=(12,4))

sns.countplot(
    data=tips,
    x="sex",
    ax=ax[0]
)

ax[0].set_title("Gender Counts")

sns.countplot(
    data=tips,
    x="day",
    ax=ax[1]
)

ax[1].set_title("Customers by Day")

plt.tight_layout()
plt.show()
```

## Dashboard 3 — Relationship Analysis

Questions:

- Which features are related?
- Are there trends?


```python
fig, ax = plt.subplots(1,2, figsize=(12,4))

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    ax=ax[0]
)

ax[0].set_title("Bill vs Tip")

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    ax=ax[1]
)

ax[1].set_title("Regression Trend")

plt.tight_layout()
plt.show()
```

## Dashboard 4 — Distribution by Category

Compare distributions across groups.


```python
fig, ax = plt.subplots(1,2, figsize=(12,4))

sns.boxplot(
    data=tips,
    x="sex",
    y="total_bill",
    ax=ax[0]
)

ax[0].set_title("Bills by Gender")

sns.violinplot(
    data=tips,
    x="sex",
    y="total_bill",
    ax=ax[1]
)

ax[1].set_title("Distribution by Gender")

plt.tight_layout()
plt.show()
```

## Dashboard 5 — Correlation Analysis

One of the most important EDA dashboards.


```python
numeric = tips.select_dtypes(include=np.number)

corr = numeric.corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

## Building a Complete EDA Dashboard

A typical dashboard combines:

- Histograms
- Boxplots
- Scatterplots
- Correlation Heatmaps


```python
fig, ax = plt.subplots(2,2, figsize=(12,8))

sns.histplot(
    tips["total_bill"],
    kde=True,
    ax=ax[0,0]
)

ax[0,0].set_title("Distribution")

sns.boxplot(
    x=tips["total_bill"],
    ax=ax[0,1]
)

ax[0,1].set_title("Outliers")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    ax=ax[1,0]
)

ax[1,0].set_title("Relationship")

sns.countplot(
    data=tips,
    x="day",
    ax=ax[1,1]
)

ax[1,1].set_title("Category Counts")

plt.tight_layout()
plt.show()
```
