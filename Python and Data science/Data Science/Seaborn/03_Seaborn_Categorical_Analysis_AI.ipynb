## What is Categorical Data?

Examples:

- Gender
- Country
- Department
- Fraud / Non-Fraud
- Purchased / Not Purchased

These are labels rather than continuous numbers.


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Why Categorical Analysis Matters

Before training a model:

- Are classes balanced?
- Which category is most common?
- Is the target variable skewed?

Categorical analysis answers these questions.


## Countplot

Countplot counts observations in each category.

### AI Usage

- Class imbalance detection
- Target analysis
- Category frequency analysis


```python
tips = sns.load_dataset("tips")

sns.countplot(
    data=tips,
    x="sex"
)

plt.title("Gender Count")
plt.show()
```

## Detecting Class Imbalance

Example:

Fraud = 100

Non-Fraud = 9900

This can cause ML problems.


```python
data = pd.DataFrame({
    "Class": ["Non-Fraud"]*90 + ["Fraud"]*10
})

sns.countplot(
    data=data,
    x="Class"
)

plt.title("Class Imbalance")
plt.show()
```

## Countplot with Hue

Hue allows grouped comparisons.


```python
tips = sns.load_dataset("tips")

sns.countplot(
    data=tips,
    x="day",
    hue="sex"
)

plt.show()
```

## Barplot

Barplot shows aggregated values.

Default:

Mean


```python
tips = sns.load_dataset("tips")

sns.barplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.title("Average Bill by Day")
plt.show()
```

## Countplot vs Barplot

Countplot → Counts observations

Barplot → Aggregates values (usually mean)


## AI Example — Model Comparison

Barplots are commonly used to compare:

- Accuracy
- Precision
- Recall
- F1 Score


```python
results = pd.DataFrame({
    "Model":["RF","XGB","SVM"],
    "Accuracy":[89,94,87]
})

sns.barplot(
    data=results,
    x="Model",
    y="Accuracy"
)

plt.title("Model Comparison")
plt.show()
```

## Catplot

Figure-level categorical visualization.

Supports:

- bar
- count
- box
- violin
- strip


```python
tips = sns.load_dataset("tips")

sns.catplot(
    data=tips,
    x="day",
    y="total_bill",
    kind="bar"
)
```

## Catplot with Hue

```python
tips = sns.load_dataset("tips")

sns.catplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex",
    kind="bar"
)
```

## Ordering Categories

Useful for professional reports.


```python
tips = sns.load_dataset("tips")

sns.countplot(
    data=tips,
    x="day",
    order=["Thur","Fri","Sat","Sun"]
)

plt.show()
```

## Using Color Palettes

```python
sns.countplot(
    data=tips,
    x="day",
    palette="deep"
)

plt.show()
```
