# Why Outliers Matter

Suppose most salaries are:

```python
[25000, 27000, 30000, 32000, 35000]
```

But one value is:

```python
1000000
```

This single value can distort:

- Mean
- Standard deviation
- Model training

Outliers can significantly affect machine learning performance.



# What is a Box Plot?

A box plot summarizes a distribution using:

- Minimum
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum

It also highlights potential outliers.


````python

import matplotlib.pyplot as plt

data = [10,12,13,15,18,20,22,25,28]

plt.boxplot(data)

plt.title("Basic Box Plot")

plt.show()
````


# Understanding Quartiles

Quartiles divide data into four equal parts.

### Q1 (25%)

25% of values lie below this point.

### Median (Q2)

Middle value of the dataset.

### Q3 (75%)

75% of values lie below this point.



# Understanding the Median

Median is often better than mean when outliers exist.

Example:

```python
[10, 11, 12, 13, 1000]
```

Mean becomes misleading.

Median remains robust.


````python

import numpy as np

data = [10,11,12,13,1000]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
````


# Interquartile Range (IQR)

Formula:

```text
IQR = Q3 - Q1
```

IQR measures the spread of the middle 50% of data.

It is commonly used for outlier detection.


````python

import numpy as np

data = [10,12,14,16,18,20,22,24]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
````


# Outlier Detection Using IQR

Rule:

```text
Lower Bound = Q1 - 1.5 * IQR

Upper Bound = Q3 + 1.5 * IQR
```

Values outside these limits are potential outliers.


````python

import numpy as np

data = [10,12,13,15,18,20,22,25,100]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Lower Bound:", lower)
print("Upper Bound:", upper)
````

# Feature Analysis with Box Plots

Use box plots to compare features.

Questions:

- Which feature has large spread?
- Which feature contains outliers?
- Which feature requires preprocessing?


````python

import matplotlib.pyplot as plt

age = [20,22,24,25,26,28,30,32,34]
salary = [25000,27000,29000,31000,33000,35000,37000,39000,100000]

plt.figure(figsize=(8,5))

plt.boxplot([age, salary])

plt.xticks([1,2], ["Age", "Salary"])

plt.title("Feature Comparison")

plt.show()
````
