```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Why Relationship Analysis Matters

Before training a model:

- Are variables related?
- Is the relationship linear?
- Is the relationship strong?
- Which features affect the target?

These questions help us build better models.


## What is Correlation?

Correlation measures how variables move together.

### Positive Correlation

As X increases, Y increases.

### Negative Correlation

As X increases, Y decreases.

### No Correlation

No visible relationship exists.


## Scatterplot

Scatterplots visualize individual observations.

One of the most important EDA tools.


```python
age = [20,25,30,35,40]
salary = [25000,35000,45000,55000,65000]

sns.scatterplot(
    x=age,
    y=salary
)

plt.title("Age vs Salary")
plt.show()
```

## Positive Correlation Example

As age increases, salary increases.

This suggests a positive relationship.


## Negative Correlation Example

```python
hours_tv = [1,2,3,4,5,6]
exam_score = [95,90,85,80,75,70]

sns.scatterplot(
    x=hours_tv,
    y=exam_score
)

plt.title("Negative Correlation")
plt.show()
```

## No Correlation Example

```python
x = np.random.rand(100)
y = np.random.rand(100)

sns.scatterplot(
    x=x,
    y=y
)

plt.title("No Correlation")
plt.show()
```

## Using DataFrames

Seaborn works directly with DataFrames.


```python
tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
```

## Hue Parameter

Hue adds a third dimension.

Useful for comparing groups.


```python
tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex"
)

plt.show()
```

## Size Parameter

Size represents an additional feature.


```python
tips = sns.load_dataset("tips")

sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    size="size"
)

plt.show()
```

## Lineplot

Used for trends over ordered data.

Common AI usage:

- Loss Curves
- Accuracy Curves
- Time Series Analysis


```python
epochs = [1,2,3,4,5]
loss = [2.5,1.8,1.2,0.8,0.5]

sns.lineplot(
    x=epochs,
    y=loss
)

plt.title("Training Loss")
plt.show()
```

## Time Series Example

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales = [120,150,180,170,220]

sns.lineplot(
    x=months,
    y=sales
)

plt.title("Monthly Sales")
plt.show()
```

## Regplot

Regression Plot

Combines:

- Scatterplot
- Regression Line

Very useful for ML intuition.


```python
tips = sns.load_dataset("tips")

sns.regplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
```

## Why Regplot Matters

Questions answered:

- Is relationship linear?
- Is trend positive?
- Is trend negative?
- Is regression appropriate?


## Feature vs Target Analysis

One of the most important ML tasks.

Questions:

Which features influence the target?


```python
experience = [1,2,3,4,5,6]
salary = [25000,30000,38000,45000,52000,65000]

sns.regplot(
    x=experience,
    y=salary
)

plt.title("Experience vs Salary")
plt.show()
```

## Outlier Detection

Scatterplots often reveal unusual observations.


```python
x = [1,2,3,4,5,6,7,8]
y = [10,12,15,18,20,22,25,100]

sns.scatterplot(
    x=x,
    y=y
)

plt.title("Outlier Example")
plt.show()
```
